```python
import matplotlib.pyplot as plt
```

```python
meses = ['Jan', 'Fev', 'Mar', 'Abr']

empresa_a = [100,120,150,180]
empresa_b = [90,130,140,170]
```

```python
plt.plot(meses, empresa_a, label='Empresa A')
plt.plot(meses, empresa_b, label="Empresa B")

plt.title("Comparação de vendas")
plt.xlabel("Mês")
plt.ylabel("Vendas")

plt.legend()
```
