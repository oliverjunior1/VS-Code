```python
import pandas as pd
```

```python
dados = {
    'nome':['joão', 'MARIA', 'João', 'pedro', 'ANA'],
    'idade':[30,None, 30,200,25],
    'cidade':['goiania', 'Goiânia', 'GOIANIA', 'Brasília', ' brasilia '],
    'salario':[3500,4200,3500,-500,None]
}
```

```python
df = pd.DataFrame(dados)
df
```

```python
df['nome'] = (df['nome'].str.lower().str.strip().str.capitalize())
df
```

```python
df['cidade'] = (df['cidade'].str.lower().str.strip().str.capitalize())
df.loc[0,'cidade'] = "Goiânia"
df.loc[2,'cidade'] = "Goiânia"
df
```

```python
df.loc[df['idade']>120, 'idade'] = None
df
```

```python
df.loc[df['salario'] < 0, 'salario'] = None
df
```

```python
df.duplicated()
```

```python
df = df.drop_duplicates()
df
```

```python
df.to_csv('dados_de_clientes_limpo.csv', index=False)
```
