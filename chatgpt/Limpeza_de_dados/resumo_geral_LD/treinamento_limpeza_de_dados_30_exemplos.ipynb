## 🟢 Iniciante 1. Valores ausentes

**Objetivo:** Identificar e preencher valores `NaN` em uma coluna numérica.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "cliente": ["Ana", "Bruno", "Carla", "Diego"],
    "idade": [25, np.nan, 31, np.nan]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
# 1) Conte os valores ausentes.
print(df["idade"].isna().sum())

# 2) Para este exercício, usamos a mediana.
mediana = df["idade"].median()

# 3) Preencha somente os valores ausentes.
df["idade"] = df["idade"].fillna(mediana)

df
```

**Comentário:** A mediana é menos sensível a valores extremos do que a média. Em um projeto real, a escolha deve depender do significado da variável.

## 🟡 Intermediário 11. Outliers pelo IQR

**Objetivo:** Detectar valores extremos usando o intervalo interquartil.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "vendas": [10, 12, 11, 13, 12, 14, 15, 11, 13, 150]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
q1 = df["vendas"].quantile(0.25)
q3 = df["vendas"].quantile(0.75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

outliers = df.loc[
    (df["vendas"] < limite_inferior) |
    (df["vendas"] > limite_superior)
]

print("Limites:", limite_inferior, limite_superior)
print("Outliers:")
print(outliers)
```

**Comentário:** IQR é uma técnica robusta, mas um outlier estatístico não significa necessariamente um erro.

## 🟡 Intermediário 12. Outliers por grupo

**Objetivo:** Detectar valores extremos considerando que grupos diferentes têm escalas diferentes.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "loja": ["A"]*5 + ["B"]*5,
    "vendas": [10, 11, 12, 9, 13, 100, 105, 98, 102, 500]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
def marcar_outlier(grupo):
    q1 = grupo["vendas"].quantile(.25)
    q3 = grupo["vendas"].quantile(.75)
    iqr = q3 - q1
    return (grupo["vendas"] < q1 - 1.5*iqr) | (grupo["vendas"] > q3 + 1.5*iqr)

df["outlier"] = df.groupby("loja", group_keys=False).apply(
    lambda g: marcar_outlier(g),
    include_groups=False
).explode().astype(bool).to_numpy()

df
```

**Comentário:** Comparar lojas juntas pode produzir falsos positivos. A distribuição de cada grupo pode ser diferente.

## 🟡 Intermediário 13. Normalização de texto

**Objetivo:** Padronizar texto removendo acentos, espaços e diferenças de caixa.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "cidade": ["São Paulo", "SÃO PAULO ", "sao paulo", " São-Paulo", "Goiania"]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
import unicodedata

def normalizar_texto(s):
    s = str(s).strip().lower()
    s = unicodedata.normalize("NFKD", s)
    s = "".join(c for c in s if not unicodedata.combining(c))
    s = s.replace("-", " ")
    s = " ".join(s.split())
    return s

df["cidade_normalizada"] = df["cidade"].map(normalizar_texto)
df
```

**Comentário:** A normalização facilita comparação, mas não deve apagar informações relevantes. Guarde a coluna original quando possível.

## 🔴 Avançado 21. Detecção de duplicidade aproximada

**Objetivo:** Encontrar nomes muito semelhantes usando distância de edição.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "cliente_id": [1, 2, 3, 4, 5],
    "nome": ["João Silva", "Joao Silva", "Maria Souza", "María Souza", "Pedro Lima"]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
# Primeiro normalize os nomes.
df["nome_norm"] = df["nome"].map(normalizar_texto)

# Distância de Levenshtein simples, implementada para treinamento.
def levenshtein(a, b):
    prev = list(range(len(b) + 1))
    for i, ca in enumerate(a, 1):
        curr = [i]
        for j, cb in enumerate(b, 1):
            curr.append(min(
                curr[-1] + 1,
                prev[j] + 1,
                prev[j-1] + (ca != cb)
            ))
        prev = curr
    return prev[-1]

pares = []
nomes = df["nome_norm"].tolist()

for i in range(len(nomes)):
    for j in range(i + 1, len(nomes)):
        d = levenshtein(nomes[i], nomes[j])
        pares.append((i, j, nomes[i], nomes[j], d))

pd.DataFrame(pares, columns=["i", "j", "nome_a", "nome_b", "distancia"])
```

**Comentário:** Fuzzy matching é útil para deduplicação, mas deve ter critérios adicionais (CPF, e-mail, endereço, data etc.) para evitar unir pessoas diferentes.

## 🔴 Avançado 23. Z-score para detecção de anomalias

**Objetivo:** Identificar observações extremas em uma distribuição aproximadamente normal.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "temperatura": [20, 21, 19, 20.5, 21.2, 19.8, 20.1, 80]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
media = df["temperatura"].mean()
desvio = df["temperatura"].std()

df["z_score"] = (df["temperatura"] - media) / desvio
df["anomalia"] = df["z_score"].abs() > 3

df
```

**Comentário:** Z-score é sensível a outliers porque média e desvio padrão também são afetados por eles. Para dados assimétricos, considere métodos robustos.

```python
mediana = df["tempo_resposta"].median()
mad = np.median(np.abs(df["tempo_resposta"] - mediana))

# 0.6745 aproxima a escala do z-score para distribuições normais.
df["z_robusto"] = 0.6745 * (df["tempo_resposta"] - mediana) / mad
df["anomalia"] = df["z_robusto"].abs() > 3.5

df
```

**Comentário:** O MAD é útil quando há valores extremos que distorceriam média e desvio padrão.

## 🔴 Avançado 26. Detecção de drift simples entre dois períodos

**Objetivo:** Comparar distribuições categóricas entre períodos para procurar mudanças suspeitas.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "periodo": ["A"]*10 + ["B"]*10,
    "canal": ["site"]*6 + ["app"]*3 + ["loja"] +
             ["site"]*2 + ["app"]*7 + ["loja"]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
tabela = pd.crosstab(df["periodo"], df["canal"], normalize="index")
print(tabela)

# Diferença absoluta entre as proporções.
if {"A", "B"}.issubset(tabela.index):
    drift = (tabela.loc["B"] - tabela.loc["A"]).abs().sort_values(ascending=False)
    print("\nMaior mudança:")
    print(drift)
```

**Comentário:** Mudanças na distribuição podem indicar mudança real de comportamento ou problema no processo de coleta.

## 🔴 Avançado 27. Tratamento de categorias desconhecidas

**Objetivo:** Preparar dados categóricos para modelos sem quebrar quando surge uma categoria nova.

## 🔴 Avançado 28. Pipeline reprodutível de limpeza

**Objetivo:** Encapsular transformações para que treino e produção recebam exatamente o mesmo tratamento.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "idade": [25, np.nan, 40, 35],
    "renda": [3000, 4500, np.nan, 5000]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

pipeline = Pipeline([
    ("imputacao", SimpleImputer(strategy="median")),
    ("escala", StandardScaler())
])

X = pipeline.fit_transform(df[["idade", "renda"]])

print("Dados originais:")
print(df)
print("\nDados após pipeline:")
print(X)
```

**Comentário:** Pipelines evitam o erro de aplicar transformações diferentes no treinamento e na produção e ajudam a prevenir data leakage.

## 🔴 Avançado 29. Data leakage na limpeza

**Objetivo:** Demonstrar como uma estatística calculada com todo o dataset pode vazar informação do conjunto de teste.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
from sklearn.model_selection import train_test_split

df = pd.DataFrame({
    "x": [10, 11, 12, 13, 14, 1000],
    "y": [0, 0, 0, 1, 1, 1]
})

treino, teste = train_test_split(df, test_size=0.33, random_state=42)
print("Treino:\n", treino)
print("\nTeste:\n", teste)
```

### 💡 Tente resolver antes de abrir a solução

```python
# ERRADO: calcular a mediana usando treino + teste.
mediana_global = df["x"].median()

# CORRETO: estatística de imputação aprendida apenas no treino.
mediana_treino = treino["x"].median()

print("Mediana global:", mediana_global)
print("Mediana treino:", mediana_treino)

# A transformação aplicada ao teste usa apenas informação aprendida no treino.
teste = teste.copy()
teste["x_limpo"] = teste["x"].fillna(mediana_treino)

teste
```

**Comentário:** Data leakage ocorre quando informações do futuro ou do conjunto de teste influenciam o processamento usado para treinar o modelo. Isso pode produzir avaliação artificialmente otimista.

## 🔴 Avançado 30. Framework de qualidade com relatório

**Objetivo:** Criar uma função reutilizável que gera métricas de qualidade de um DataFrame.

**Passo a passo:**
1. Crie/inspecione os dados.
2. Identifique o problema de qualidade.
3. Aplique uma transformação controlada.
4. Valide o resultado.

### 🧪 Dados propositalmente sujos


```python
df = pd.DataFrame({
    "id": [1, 2, 2, 4, None],
    "nome": ["Ana", "Bruno", "Bruno", None, "Eva"],
    "idade": [25, 30, -5, 40, 200],
    "email": ["a@x.com", "b@x.com", "b@x.com", "sem-email", None]
})
df
```

### 💡 Tente resolver antes de abrir a solução

```python
def relatorio_qualidade(df):
    relatorio = pd.DataFrame({
        "tipo": df.dtypes.astype(str),
        "nulos": df.isna().sum(),
        "pct_nulos": (df.isna().mean() * 100).round(2),
        "unicos": df.nunique(dropna=True),
        "duplicados_linha": df.duplicated().sum()
    })

    return relatorio

relatorio = relatorio_qualidade(df)
print(relatorio)

# Regras adicionais de negócio:
print("\nIdades inválidas:", (~df["idade"].between(0, 120)).sum())
print("IDs nulos:", df["id"].isna().sum())
print("E-mails suspeitos:",
      (~df["email"].fillna("").str.contains("@", regex=False)).sum())
```

**Comentário:** Um bom processo de qualidade não é apenas uma sequência de `dropna()` e `drop_duplicates()`: ele deve produzir métricas, regras, rastreabilidade e critérios claros de aceitação.
