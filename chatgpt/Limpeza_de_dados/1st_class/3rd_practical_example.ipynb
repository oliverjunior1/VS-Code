```python
import pandas as pd

dados = {
    "nome": [
        "João",
        "Maria",
        "Pedro",
        "João",
        "Ana"
    ],
    
    "idade": [
        32,
        28,
        None,
        32,
        29
    ],
    
    "salario": [
        3500,
        4200,
        3800,
        3500,
        None
    ]
}

df = pd.DataFrame(dados)

print("DADOS ORIGINAIS")
print(df)

print("\nVALORES AUSENTES")
print(df.isnull().sum())

print("\nDUPLICADOS")
print(df.duplicated().sum())

```

# Remover duplicates

```python
df = df.drop_duplicates()
df
```

```python
media_idade = df['idade'].mean()
media_idade
```

```python
df['idade'] = df['idade'].fillna(media_idade)
df
```

# Preencher salário com a média

```python
media_salario = df['salario'].mean()
media_salario
```

```python
df['salario'] = df['salario'].fillna(media_salario)
df
```
