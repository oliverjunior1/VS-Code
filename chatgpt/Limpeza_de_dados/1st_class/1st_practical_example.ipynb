```python
import pandas as pd

dados = {
    "nome": [
        "João",
        "Maria",
        "Pedro",
        "Ana",
        "Carlos"
    ],
    
    "idade": [
        32,
        28,
        None,
        29,
        None
    ],
    
    "salario": [
        3500,
        4200,
        3800,
        None,
        5000
    ]
}

df = pd.DataFrame(dados)

print("Dados:")
print(df)

print("\nValores ausentes:")
print(df.isnull().sum())
```

### Removendo valores ausentes
