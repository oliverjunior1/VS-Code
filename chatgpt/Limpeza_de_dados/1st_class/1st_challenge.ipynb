```python
import pandas as pd

dados = {
    "nome": [
        "João",
        "Maria",
        "Pedro",
        "João",
        "Ana",
        "Carlos"
    ],
    
    "idade": [
        32,
        None,
        25,
        32,
        None,
        41
    ],
    
    "salario": [
        3500,
        4200,
        None,
        3500,
        2800,
        None
    ]
}

df = pd.DataFrame(dados)
```

### 1️⃣ Descobrir quantos valores ausentes existem.

```python
sum_of_null =df.isnull().sum()
sum_of_null
```

### 2️⃣ Descobrir quantas linhas duplicadas existem.

```python
duplicados = df.duplicated().sum()
duplicados
```

### 3️⃣ Remover os duplicados.

```python
df = df.drop_duplicates()
df
```

### 4️⃣ Substituir os valores ausentes de ``idade`` pela média.

```python
media_idades = df['idade'].mean()
df['idade'] = df['idade'].fillna(media_idades)
df
```
