```python
import matplotlib.pyplot as plt
```

```python
produtos = ["Notebook", "Mouse", "Teclado", "Monitor", "Headset"]
quantidades = [35,120,85,50,95]
```

```python
plt.bar(produtos, quantidades)

plt.title("Quantidade vendida por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade")

plt.grid(axis='y')
```

```python
plt.barh(produtos, quantidades)

plt.title("Quantidade vendida por produto")
plt.xlabel("Quantidade")
plt.ylabel('Produto')
```
