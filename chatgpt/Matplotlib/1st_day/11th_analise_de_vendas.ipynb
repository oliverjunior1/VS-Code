```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dados = {
    'produto':['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'vendas':[12000,5000,7000,15000,9000]
}
```

```python
df = pd.DataFrame(dados)
df
```

# Primeiro: descobrir o maior faturamento

```python
print(df['vendas'].max())
```

# Segundo o gráfico

```python
plt.bar(
    df['produto'], df['vendas']
)

plt.title('Faturamento por produto')
plt.xlabel('Produto')
plt.ylabel('Faturamento')

plt.xticks(rotation= 45)
```

# Ordenar os dados

```python
df = df.sort_values(
    'vendas',
    ascending=False
)

df
```
