```python
import pandas as pd

dados = {
    "nome": ["João", "Maria", "Pedro", "João", "Ana", "Carlos"],
    "idade": [32, 28, None, 32, 29, 40]
}

df = pd.DataFrame(dados)

print("Antes:")
print(df)

media_idade = df["idade"].mean()

df["idade"] = df["idade"].fillna(media_idade)

print("\nDepois:")
print(df)
```
