### Importar os pacotes:

```python
import pandas as pd
import numpy as np
```

### Criar o dataframe

```python
df = pd.DataFrame({
    'nome': ['Ana', 'Pedro', np.nan, 'Maria'],
    'idade': [20, np.nan, 35, 40]
})
```

# 1. Mostrar quantidade de nulos por coluna.

```python
df.isnull().sum()
```

# 2. Mostrar percentagem de nulos.

```python
media = (df.isnull().sum().sum()/df.size)*100
media
```

# 3. Substituir idade nula pela média

```python
df['idade'] = df['idade'].fillna(df['idade'].mean())
df['idade']
```

# 4. Remover linhas onde nome é nulo 

```python
df = df.dropna(subset=['nome'])
df['nome']
```
