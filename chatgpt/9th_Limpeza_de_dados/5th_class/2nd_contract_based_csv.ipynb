```python
from io import StringIO
import pandas as pd
```

```python
text = "id;region;qty\n001;NA;2\n002;EU;\n"
```

```python

df = pd.read_csv(StringIO(text), sep=";", dtype="string",
                 keep_default_na=False)
```

```python
df.to_string(index=False)
```

```python
assert df.loc[0, 'id'] == "001"
```

```python
assert df.loc[0, 'region'] == "NA"
```
