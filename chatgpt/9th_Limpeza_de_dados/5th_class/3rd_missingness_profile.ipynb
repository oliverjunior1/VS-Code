```python
import pandas as pd
```

```python
raw = pd.DataFrame({"id":["a", "b", "c"], "qty":[1, None, 3]})
```

```python
profile = pd.DataFrame({"missing":raw.isna().sum(),"missing_rate":raw.isna().mean()})
```

```python
profile.to_string()
```

```python
assert profile.loc["qty", "missing"] == 1
```
