```python
import pandas as pd
```

```python
dados = {
    "nome":["Ana", "João", "Maria", "Pedro", "Carlos"],
    "idade":[25,40,32,28,50],
    "salario":[3500,5000,4500,2800,7000]
}
```

```python
df = pd.DataFrame(dados)
df
```

```python
clientes_altos_salarios = df[df["salario"] > 4000]
clientes_altos_salarios
```
