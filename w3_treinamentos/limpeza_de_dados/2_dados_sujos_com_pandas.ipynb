### Importa a biblioteca pandas, usada para trabalhar com tabelas de dados

```python
import pandas as pd
```

#### Cria uma tabela com dados sujos para praticar limpeza

```python
dados = {
    "nome": [" Ana ", "Bruno", "CARLA", "daniel", None],
    "idade": [25, None, 31, -5, 40],
    "cidade": ["São Paulo", "rio de janeiro", "CURITIBA", None, "salvador"],
    "email": ["ana@email.com", "bruno@email.com", None, "daniel@email.com", "sem_email"]
}
```

#### Transforma o dicionário em um DataFrame, que é uma tabela do pandas


```python
df = pd.DataFrame(dados)
```

### Mostra a tabela original antes da limpeza

```python
df
```

### Remove espaços extras no começo e no fim dos nomes

```python
df['nome'] = df['nome'].str.strip()
df
```

### Padroniza os nomes para começar com letra maiúscula

```python
df['nome'] = df['nome'].str.title()
df
```

### Preenche nomes vazios com o texto "Nome não informado"

```python
df['nome'] = df['nome'].fillna("Nome não informado")
df
```
