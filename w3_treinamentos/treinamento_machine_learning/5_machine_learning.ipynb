### Treino 2: Medindo O Erro Do Modelo

### Importa pandas para trabalhar com tabelas

```python
import pandas as pd
```

### Importa o modelo de regressão linear


```python
from sklearn.linear_model import LinearRegression
```

### Importa funções para medir o erro do modelo


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

### Cria os dados de treino


```python
dados = {
    "horas_estudo": [1, 2, 3, 4, 5, 6, 7, 8],
    "nota": [50, 55, 65, 70, 75, 85, 90, 95]
}
```

### Transforma os dados em uma tabela

```python
df = pd.DataFrame(dados)
```

### Mostra a tabela usada no treino


```python
print("DADOS")
print(df)
```

### Separa a entrada do modelo
### X precisa ser uma tabela, por isso usamos dois colchetes

```python
X = df[['horas_estudo']]
```

### Separa a resposta correta que o modelo deve aprender


```python
y = df['nota']
```

### Cria o modelo de regressão linear


```python
modelo = LinearRegression()
```

### Treina o modelo com os dados


```python
modelo.fit(X, y)
```

### Faz previsões para os próprios dados de treino

```python
previsoes = modelo.predict(X)
```

### Cria uma nova coluna com as previsões feitas pelo modelo


### O erro é a diferença entre a nota real e a nota prevista

```python
df['nota_prevista'] = previsoes.round(1)
```

### Calcula o erro absoluto médio
### Ele mostra, em média, quantos pontos o modelo erra

```python
mae = mean_absolute_error(df['nota'], df['nota_prevista'])
```

### Calcula o erro quadrático médio
### Ele pune erros grandes com mais força

```python
mse = mean_squared_error(df['nota'], df['nota_prevista'])
```

### Mostra a tabela com previsão e erro

```python
print("\nTABELA COM PREVISÕES")
print(df)
```

### Mostra as métricas do modelo

```python
print("\nMÉTRICAS DO MODELO")
print(f"Erro absoluto médio: {mae:.2f} pontos")
print(f"Erro quadrático médio: {mse:.2f}")
```

### Cria um novo aluno para previsão


```python
novo_aluno = pd.DataFrame({
    'horas_estudo:[9]'
})
```

### Faz a previsão para o novo aluno

```python
previsao_novo_aluno = modelo.predict(novo_aluno)
```
