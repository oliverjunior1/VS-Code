```python
# Importa pandas para trabalhar com tabelas
import pandas as pd

# Cria uma tabela com dados sujos
dados = {
    "cliente": ["Ana", "Bruno", "Ana", "Carla", "Joaquim", "Daniel", "Bruno", None, "Alyne", "Joaquim"],
    "idade": [25, 30, 25, None, 43, -8, 30, 41, -12, 43],
    "cidade": ["São Paulo", "RIO DE JANEIRO", "São Paulo", "curitiba", "Anápolis", None, "RIO DE JANEIRO", "salvador", "Goiania", "Anápolis"],
    "compra": [120.50, 80.00, 120.50, None, -50.00, 200.00, 80.00, -15.00, 150.00, -50.00],
    "status": ["ativo", None, "ativo", "inativo", None, "ativo", None, "ativo", None, None]
}

# Transforma o dicionário em uma tabela do pandas
df = pd.DataFrame(dados)

# Mostra os dados antes da limpeza
print("DADOS ORIGINAIS")
print(df)

# Conta quantas linhas duplicadas existem na tabela
duplicadas = df.duplicated().sum()

# Conta quantos valores vazios existem em cada coluna
valores_vazios = df.isna().sum()

# Remove linhas duplicadas da tabela
df = df.drop_duplicates()

# Preenche cliente vazio com um texto padrão
df["cliente"] = df["cliente"].fillna("Cliente não informado")

# Padroniza as cidades para ficarem com a primeira letra maiúscula
df["cidade"] = df["cidade"].fillna("Cidade não informada")
df["cidade"] = df["cidade"].str.title()

# Calcula a média das idades válidas
media_idade = df[df["idade"] > 0]["idade"].mean()

# Preenche idade vazia com a média
df["idade"] = df["idade"].fillna(media_idade)

# Substitui idade negativa pela média
df.loc[df["idade"] < 0, "idade"] = media_idade

# Arredonda idade e transforma em número inteiro
df["idade"] = df["idade"].round(0).astype(int)

# Calcula a média das compras válidas
media_compra = df[df["compra"] > 0]["compra"].mean()

# Preenche compra vazia com a média
df["compra"] = df["compra"].fillna(media_compra)

# Substitui compra negativa pela média
df.loc[df["compra"] < 0, "compra"] = media_compra

# Arredonda compra para 2 casas decimais
df["compra"] = df["compra"].round(2)
df["status"] = df["status"].fillna("ativo")

# Mostra os dados depois da limpeza
print("\nDADOS LIMPOS")
print(df)

# Mostra um pequeno relatório de qualidade
print("\nRELATÓRIO DE QUALIDADE")
print(f"Linhas duplicadas encontradas: {duplicadas}")
print("Valores vazios antes da limpeza:")
print(valores_vazios)
print(f"Total de clientes após limpeza: {len(df)}")
print(f"Média de idade após limpeza: {df['idade'].mean():.1f}")
print(f"Média de compra após limpeza: R$ {df['compra'].mean():.2f}")
```
